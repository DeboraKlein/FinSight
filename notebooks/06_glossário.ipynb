Glossário do Projeto (Termos-Chave)

| Termo | Categoria | Definição | Fonte(s) |
| :--- | :--- | :--- | :--- |
| Estresse Agudo |Risco/Negócio | Fase de risco do cliente que antecede a migração para a inadimplência ($\mathbf{< 30 \text{ dias}}$ de delay). O modelo deve agir na janela crítica 0 a 14 dias para evitar negativação. | N01, N04 |
| RAROC | Finanças | Retorno Ajustado ao Risco (Risk-Adjusted Return on Capital). Métrica-chave que mede a rentabilidade do projeto dividida pelo capital alocado para cobrir o risco. O projeto finalizou com um RAROC Otimizado de $\mathbf{3640.00\%}$ a.a.. | N02, N05 | 
| PECLD | Finanças/Contábil | Perda Esperada em Créditos de Liquidação Duvidosa. Provisão prospectiva exigida pelo IFRS 9. O projeto visa reduzir a PECLD ao evitar a migração do cliente para o Estágio 2. | N02 |
| IFRS 9 (Estágios) | Regulatório | Norma contábil internacional. O projeto foca em evitar que o cliente migre do Estágio 1 (risco baixo/normal) para o Estágio 2 (risco significativamente aumentado). | N02 |
| Threshold Ótimo | ML/Estratégia | O ponto de corte ideal (Y_PROBA $\ge$ threshold) encontrado para maximizar o RAROC. O valor final de 0.10 garante máxima seletividade e lucro. | N03, N05 |
| Taxa de Cura | Risco/Negócio | Percentual de clientes que aceitaram a oferta de consolidação e tiveram sucesso, ou seja, não entraram em default (Y_Target = 0). | N05 |
| PCR Ideal | Estratégia/Oferta | Percentual de Comprometimento de Renda Ideal. O modelo V2 deve modular o PCR de cada oferta, garantindo que o novo crédito seja sustentável para o cliente. | N01, N02 |
| Pipeline de Inferência | Operacionalização | Estrutura de código sequencial e automatizada responsável por receber os dados brutos e aplicar todas as etapas de feature engineering, scoring e decisão, culminando na oferta. Essencial para o deploy em produção. | N04 |
| Processamento em Lote (Batch) | Operacionalização | Forma de processar grandes volumes de dados de uma só vez (ex: pontuar toda a carteira diariamente) para identificar clientes para a Intervenção Proativa. | N04 |
| Backtesting (Pós-Mortem) | Monitoramento | Simulação da performance da estratégia em dados históricos (fora do período de treinamento) para calcular o RAROC Realizado e validar as premissas financeiras. | N05 |
| Monitoramento de Drift | Monitoramento | Acompanhamento contínuo da estabilidade das Features e da performance do modelo em produção. É vital para garantir que o RAROC Realizado permaneça próximo do RAROC Otimizado. | N05 |
| Intervenção Proativa | Estratégia | Ação de oferecer um Empréstimo de Consolidação ao cliente em Estresse Agudo, antes que ele atinja o default ou migre para linhas de crédito de altíssimo custo. | N01, N02, N06 |
| Target de ML (Estresse) | ML | A variável binária (0 ou 1) que o modelo de ML está tentando prever. No projeto, é o cliente que apresenta Estresse Agudo (< 30 dias de atraso). | N01b, N03, N05 |
